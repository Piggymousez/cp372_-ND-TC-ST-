```python
import pandas as pd

# โหลดไฟล์ CSV (เปลี่ยนชื่อไฟล์ตามที่อัปโหลด)
df = pd.read_csv('Chocolate Sales.csv')

# ดูข้อมูลเบื้องต้น
df.head()
```

```python
# ดูขนาดและชนิดของข้อมูล
df.info()

# ดูสถิติเบื้องต้น
df.describe(include='all')
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
# เปลี่ยนชื่อคอลัมน์ให้อ่านง่าย 1.Strip ลบช่องว่างหน้าหลัง, 2.Lower เปลี่ยนเป็นตัวพิม 3.Replace แทนที่ช่องว่างด้วยขีดล่าง
# Example (1,2)
# 1.จาก Sales Person เป็น sales_person
# 2.จาก Boxes Shipped เป็น boxes_shipped
```

```python
# ลบช่องว่างหน้า/หลังในทุก cell ที่เป็น string
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

# เช็คค่าว่าง
df.isnull().sum()
```

```python
# ลบ $ และ , แล้วแปลงเป็น float
df['amount'] = df['amount'].replace('[\$,]', '', regex=True).astype(float)
```

```python
# แปลงคอลัมน์ date เป็น datetime
df['date'] = pd.to_datetime(df['date'], format='%d-%b-%y')
```

```python
# สร้างฟิลด์ใหม่ คือ
def get_festivals(date): #ฟังก์ชั่นนี้สร้างขึ้นมาเพื่อรับ วันที่ (date) หนึ่งวัน แล้วดูว่าวันนั้นตรงกับเทศกาลไหนบ้างละคืนค่าชื่อเทศกาลนั้น ๆ กลับไปในรูปแบบข้อความ
    month = date.month
    day = date.day
    festivals = [] #สร้างลิสต์ว่างไว้เก็บชื่อเทศกาลที่ตรงกับวันนั้น

    if (month == 12 and day >= 25) or (month == 1 and day <= 3): #ช่วง ปีใหม่ (25 ธ.ค.–3 ม.ค.)
        festivals.append("New Year")
    if (month == 2 and 10 <= day <= 16): #ช่วง วาเลนไทน์ (10–16 ก.พ.)
        festivals.append("Valentine's Day")
    if (month == 3 and 10 <= day <= 15): #ช่วง ไวท์เดย์ (10–15 มี.ค.)
        festivals.append("White Day")
    if (month == 4): #ตลอดเดือน เมษายน ถือว่าเป็นช่วง Easter
        festivals.append("Easter")
    if (month == 5 and 7 <= day <= 14): #ช่วง Mother's Day (7–14 พ.ค.)
        festivals.append("Mother's Day")

    return ', '.join(festivals) if festivals else "None Festival" #ถ้าช่วง festival อยู่ในเงื่อนไขจะใส่ชื่อเทศกาลนั้นลงไป แต่ถ้าไม่มีจะคืนค่า "None festival"

# ใช้ฟังก์ชันใหม่
df['festival'] = df['date'].apply(get_festivals)
```

```python
# ดูค่าที่ไม่ซ้ำใน country
df['country'].unique()

# ดูค่าที่ไม่ซ้ำใน product
df['product'].unique()
```

```python
df.to_csv('cleaned_chocolate_sales.csv', index=False)

# ดาวน์โหลดไฟล์กลับเครื่อง
files.download('cleaned_chocolate_sales.csv')
```
